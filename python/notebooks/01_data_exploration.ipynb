```python
import numpy as np
import pandas as pd
```

```python
rng = np.random.default_rng(42)
n = 800  # ~6.5 hours if one reading every 30s

healthy = pd.DataFrame({
    "latency_ms":     rng.normal(180, 25, n).clip(80, 350),
    "error_rate_pct": rng.normal(0.2, 0.08, n).clip(0, 1.5),
    "cpu_pct":        rng.normal(22, 5, n).clip(5, 45),
    "memory_pct":     rng.normal(40, 4, n).clip(25, 55),
    "db_connections": rng.normal(20, 3, n).clip(5, 40),
    "label": 0,  # 0 = healthy (only for YOUR evaluation, not for training)
})
healthy.to_csv("../data/healthy.csv", index=False)
```

```python
faults = healthy.copy()

faults.loc[0:30, "latency_ms"] = 800
faults.loc[0:30, "db_connections"] = 70

faults.loc[31:60, "error_rate_pct"] = 20

faults.loc[61:90, "memory_pct"] = 90

faults.loc[91:120, "latency_ms"] = 1200
faults.loc[91:120, "error_rate_pct"] = 15

faults.loc[121:150, "cpu_pct"] = 90
faults.loc[121:150, "memory_pct"] = 85

faults.loc[151:180, "cpu_pct"] = 95
faults.loc[151:180, "cost_usd_hour"] = 0.30

faults.to_csv("../data/faults.csv", index=False)

print("Fault dataset created:", faults.shape)
```
